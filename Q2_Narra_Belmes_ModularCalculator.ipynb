### What functions did you create in your calculator program?
*I created four functions: `add_numbers()`,`subtract_numbers()`, `multiply_numbers()` and `divide_numbers()`. All of them perform basic arithmetic operations.*

### What parameters did your functions use?
*My functions all use two float parameters, `num1` and `num2`.*

### What arguments were passed when the functions were called?
*The arguments are two float inputs from the user, also named `num1` and `num2`. For my example, I inputted 10 and 5, so those are the arguments that were passed to the functions.*

### How did your program use the returned value?
*My program used the value returned from the current function (the chosen one of the four arithmetic functions) by outputting it back to the user as the result of the operation.*

### Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
*It is better to divide the program into functions because this allows us to easily navigate, test and update the program. We can also observe clear separation of concerns within the program, which improves readability to other users.
This allows us to work better—instead of taking on the entire problem at once and feeling overwhelmed, we can decompose it into smaller programs and work on each part until we've finally written the full program in modular stages.*




```python
# This is a simple modular mini-calculator program that serves as my FA output.
# It asks the user to provide 2 numbers and select an operation, then computes the result.

# Calculates the sum of two quantities.
def add_numbers(num1, num2):
  return num1 + num2

# Calculates the difference of two quantities.
def subtract_numbers(num1, num2):
  return num1 - num2

# Calculates the product of two quantities.
def multiply_numbers(num1, num2):
  return num1 * num2

# Calculates the quotient of two quantities. This contains a safeguard against division by zero.
def divide_numbers(num1, num2):
  if num2 == 0:
    return "Cannot divide by zero."
  return num1 / num2

# Prompts the user for the first and second numbers. (num1, num2)
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\nChoose operation:\n1 - Addition\n2 - Subtraction\n3 - Multiplication\n4 - Division")

# Prompts the user for their choice. (1-4)
# Then, performs the appropriate operation through an if-elif selection structure.
choice = int(input("\nEnter choice: "))
result = None

if choice == 1:
  result = add_numbers(num1, num2)
elif choice == 2:
  result = subtract_numbers(num1, num2)
elif choice == 3:
  result = multiply_numbers(num1, num2)
elif choice == 4:
  result = divide_numbers(num1, num2)

print("\nResult:", result)
```
